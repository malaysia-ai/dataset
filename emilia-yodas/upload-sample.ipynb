```python
from glob import glob
import os

repository = 'malaysia-ai/Emilia-YODAS-Voice-Conversion'
folder = 'Emilia-YODAS_permutate'
files = glob(f'{folder}/**/*.json', recursive = True)
len(files)
```

```python
from multiprocess import Pool
import itertools
import json

def chunks(l, n):
    for i in range(0, len(l), n):
        yield (l[i: i + n], i // n)

def multiprocessing(strings, function, cores=6, returned=True):
    df_split = chunks(strings, len(strings) // cores)
    pool = Pool(cores)
    pooled = pool.map(function, df_split)
    pool.close()
    pool.join()

    if returned:
        return list(itertools.chain(*pooled))
```

```python
import json
from collections import defaultdict
import random
from tqdm import tqdm

def loop(files):
    files, index = files
    samples = []
    for f in tqdm(files):
        with open(f) as fopen:
            d = json.load(fopen)

        speakers = defaultdict(list)
        for d_ in d:
            speakers[d_['reference_audio']].append(d_)
        for k, v in speakers.items():
            samples.extend(random.sample(v, min(len(v), 3)))
    return samples
```

```python
samples = multiprocessing(files, loop, cores = 30)
```

```python
len(samples)
```

```python
audios = set()
for s in tqdm(samples):
    audios.add(s['reference_audio'])
    audios.add(s['target_audio'])

len(audios)
```

```python
with open('emilia-audio.json', 'w') as fopen:
    json.dump(list(audios), fopen)
```
