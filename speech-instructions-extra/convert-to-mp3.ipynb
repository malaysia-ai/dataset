```python
from glob import glob
import soundfile as sf
from tqdm import tqdm
import mp
```

```python
!du -hs ASR-PART3-Train-audio
```

```python
files = glob('*/*.mp3')
len(files)
```

```python
def loop(files):
    files, _ = files
    for f in tqdm(files):
        y, sr = sf.read(f)
        sf.write(f, y, sr)
```

```python
mp.multiprocessing(files, loop, cores = 20, returned = False)
```
