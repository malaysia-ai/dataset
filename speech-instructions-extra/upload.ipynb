```python
from glob import glob
from tqdm import tqdm
import json
```

```python
alls = []
for f in tqdm(glob('*/*.parquet')):
    try:
        with open(f) as fopen:
            d = json.load(fopen)
        for d_ in d:
            d_['start'] = None
            d_['end'] = None
            d_['context'] = None
            d_['system'] = None
            d_['sliced_audio_filename'] = None
            alls.append(d_)
    except:
        pass
```
